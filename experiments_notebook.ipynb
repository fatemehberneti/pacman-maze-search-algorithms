# one maze compare time [ Tree ]

```python
import matplotlib.pyplot as plt
from pacmanCompareOneT import *

# Reconstruct the path from parent mapping
def reconstruct_path(parent, goal):
    path = []
    current = goal
    while current is not None:
        path.append(current)
        current = parent[current]
    path.reverse()  # Reverse to get path from start to goal
    return path

# Measure the runtime of each algorithm
def measure_runtime(algorithm, start, goal):
    start_time = time.time()
    algorithm(start, goal)
    end_time = time.time()
    return end_time - start_time

# Compare the algorithms
bfs_time = measure_runtime(bfs_tree, start, goal)
dfs_time = measure_runtime(dfs_tree, start, goal)
ids_time = measure_runtime(ids_tree, start, goal)
manhattan_time = measure_runtime(a_star_tree_manhattan, start, goal)
euclidean_time = measure_runtime(a_star_tree_euclidean, start, goal)

# Plot the runtime comparison
algorithms = ['BFS', 'DFS', 'IDS', 'MANHATTAN', 'EUCLIDEAN']
times = [bfs_time, dfs_time, ids_time, manhattan_time, euclidean_time]

plt.bar(algorithms, times, color=['blue', 'green', 'red', 'purple', 'pink'])
plt.ylabel('Time (seconds)')
plt.title('Comparison of Algorithms Runtimes')
plt.show()
```
